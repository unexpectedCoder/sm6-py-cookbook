## Копирование

В языке Python всё является объектом.
Чтобы не замедлять работу интерпретатора принято следующее соглашение: сложные объекты по умолчанию не копируются (не клонируются).
Вместо этого выполняются операции со ссылками на объекты.

В чём это выражается в контексте работы с NumPy?
Создадим один массив и затем присвоим его другой переменной:

```python
a = np.arange(10)
b = a
print("a =", a)
print("b =", b)
```

Вроде бы, получили то, что ожидали.
**Но!**
Изменим один из элементов массива `a`:

```python
a[1] = -7
```

Что увидим при выводе?
А вот что:

```python
print("a =", a)
print("b =", b)
```

Изменился и массив `b`.
И это совсем не удивительно, ведь на самом деле в `b` хранится не сам массив, а лишь ссылка на массив `a`.
Изменился массив `a`, а ссылка `b` на него (его адрес в памяти компьютера) осталась той же.
Всегда будьте внимательны, когда приравниваете сложные объекты друг к другу.

Ситуация ещё интереснее, поскольку мы можем изменить массив `a`, используя ссылку `b`:

```python
b[0] = 177
print("a =", a)
print("b =", b)
```

Как правило, такие махинации заканчиваются плохо, ведь код запутывается.

Если вы хотите иметь массив `b` такой же, как `a`, но независимый от него, то на помощь приходит метод `np.copy()`:

```python
a = np.arange(10)
b = a.copy()

a[1] = -7
b[0] = 177

print("a =", a)
print("b =", b)
```

Массивы изменяются независимо.

## Арифметические операции

Массивы NumPy поддерживают все арифметические операции Python и стандартной библиотеки `math` и даже больше!

Заведём пару массивов **одинаковой формы** и рассмотрим некоторые операции.

```python
a = np.array([1., 2, 3, 4, 5])
b = np.array([-1., 0, -5, -11, 7])
```

```python
a + 1
```

```python
2 * a
```

```python
a / 3
```

```python
a ** 2
```

```python
a ** 0.5
```

```python
a + b
```

```python
a - b
```

```python
a * b
```

```python
a / (2 + b)
```

```python
a ** b
```

## Математика посложнее

```python
a = np.arange(-np.pi, np.pi + 1e-6, 2 * np.pi / 3)
np.degrees(a)
```

```python
# тригонометрическим функциям аргумент нужен
# обязательно в радианах
print("sin a =", np.sin(a))
print("cos a =", np.cos(a))
print("tan a =", np.tan(a))
```

```python
np.exp(a)
```

```python
b = np.random.randint(-10, 11, 5)
b
```

```python
print("min b =", b.min())
print("max b =", b.max())
```

```python
# индексы мин. и макс. элементов
print("argmin b =", b.argmin())
print("argmax b =", b.argmax())
```

```python
# сумма элементов
b.sum()
```

Стоит отметить, что функции sin, cos и прочие являются *универсальными функциями* NumPy.
Универсальны они потому, что позволяют работать как с массивами, так и обычными числами.

## Операции с двумерными массивами

В NumPy имеется понятие *оси (axis)*.
Ось 0 обозначает строки, ось 1 означает столбцы и т.д.

```python
a = np.arange(12).reshape((3, 4))
a
```

```python
# сумма элементов по строкам,
# т.е. суммируются элементы в столбцах
a.sum(axis=0)
```

```python
# аналогично сумма элементов по столбцам
a.sum(axis=1)
```

```python
print("min(axis=0) =", a.min(axis=0))
print("min(axis=1) =", a.min(axis=1))
```
